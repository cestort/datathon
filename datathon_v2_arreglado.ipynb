# Escenario 1

```python
# =========================
# Guardamos una copia CRUDA de x8 (para dummies más adelante)
x8_train_raw = X_train['x8'].copy()
x8_test_raw  = X_test['x8'].copy()



# =========================
# Escenario 1 — crudo (sin x9)
# =========================

esc1_train = X_train.copy()
esc1_test  = X_test.copy()

esc1_train = X_train.drop(columns=['x9']).copy()
esc1_test  = X_test.drop(columns=['x9']).copy()
```

# Escenario 2

```python

# =========================
# Escenario 2 — imputación + Yeo-Johnson (solo continuas; NO tocar x8)
# =========================
# ============= ESCENARIO 2 (IMPUTACIÓN + YEO-JOHNSON) =============
# Nota: x8 (categórica) no se toca en Esc2.



esc2_train = esc1_train.copy()
esc2_test  = esc1_test.copy()

# ---------- IMPUTACIÓN MEDIANA (continuas) ----------
# x1
_imp_x1 = SimpleImputer(strategy='median')
esc2_train['x1'] = _imp_x1.fit_transform(esc2_train[['x1']]).ravel()
esc2_test['x1']  = _imp_x1.transform(esc2_test[['x1']]).ravel()

# x2
_imp_x2 = SimpleImputer(strategy='median')
esc2_train['x2'] = _imp_x2.fit_transform(esc2_train[['x2']]).ravel()
esc2_test['x2']  = _imp_x2.transform(esc2_test[['x2']]).ravel()

# x3
_imp_x3 = SimpleImputer(strategy='median')
esc2_train['x3'] = _imp_x3.fit_transform(esc2_train[['x3']]).ravel()
esc2_test['x3']  = _imp_x3.transform(esc2_test[['x3']]).ravel()

# x4
_imp_x4 = SimpleImputer(strategy='median')
esc2_train['x4'] = _imp_x4.fit_transform(esc2_train[['x4']]).ravel()
esc2_test['x4']  = _imp_x4.transform(esc2_test[['x4']]).ravel()

# x5
_imp_x5 = SimpleImputer(strategy='median')
esc2_train['x5'] = _imp_x5.fit_transform(esc2_train[['x5']]).ravel()
esc2_test['x5']  = _imp_x5.transform(esc2_test[['x5']]).ravel()

# x6
_imp_x6 = SimpleImputer(strategy='median')
esc2_train['x6'] = _imp_x6.fit_transform(esc2_train[['x6']]).ravel()
esc2_test['x6']  = _imp_x6.transform(esc2_test[['x6']]).ravel()

# x7
_imp_x7 = SimpleImputer(strategy='median')
esc2_train['x7'] = _imp_x7.fit_transform(esc2_train[['x7']]).ravel()
esc2_test['x7']  = _imp_x7.transform(esc2_test[['x7']]).ravel()

# x10
_imp_x10 = SimpleImputer(strategy='median')
esc2_train['x10'] = _imp_x10.fit_transform(esc2_train[['x10']]).ravel()
esc2_test['x10']  = _imp_x10.transform(esc2_test[['x10']]).ravel()

# ---------- YEO-JOHNSON SOLO EN x1, x2, x3, x5 ----------
# Usamos la funcion FitTrnasformaColumn para AJUSTAR en TRAIN (devuelve columna transformada + lambda)


# x1
esc2_train['x1'], lam_x1 = FitTransformColumn(esc2_train['x1'].values, method='yeojohnson')
esc2_test['x1'] = yeojohnson(esc2_test['x1'].values, lmbda=lam_x1)

# x2
esc2_train['x2'], lam_x2 = FitTransformColumn(esc2_train['x2'].values, method='yeojohnson')
esc2_test['x2'] = yeojohnson(esc2_test['x2'].values, lmbda=lam_x2)

# x3
esc2_train['x3'], lam_x3 = FitTransformColumn(esc2_train['x3'].values, method='yeojohnson')
esc2_test['x3'] = yeojohnson(esc2_test['x3'].values, lmbda=lam_x3)

# x5
esc2_train['x5'], lam_x5 = FitTransformColumn(esc2_train['x5'].values, method='yeojohnson')
esc2_test['x5'] = yeojohnson(esc2_test['x5'].values, lmbda=lam_x5)

# ---------- Comprobación rápida ----------
_cols_cont = ['x1','x2','x3','x4','x5','x6','x7','x10']
assert not np.isnan(esc2_train[_cols_cont].values).any(), "NaN en continuas TRAIN (Esc2)"
assert not np.isnan(esc2_test[_cols_cont].values).any(),  "NaN en continuas TEST  (Esc2)"
print("Escenario 2 OK: continuas imputadas y transformadas; x8 intacta.")
print(f"Lambdas YJ → x1:{lam_x1:.4f} | x2:{lam_x2:.4f} | x3:{lam_x3:.4f} | x5:{lam_x5:.4f}")


```

# Escenario 3 Escalado

```python
# ============= ESCENARIO 3 (ESCALADO) =============
from sklearn.preprocessing import StandardScaler
import numpy as np

# Partimos del Escenario 2 (ya imputado + YJ en x1,x2,x3,x5)
esc3_train = esc2_train.copy()
esc3_test  = esc2_test.copy()

# Creamos un scaler POR COLUMNA para máxima trazabilidad
sc_x1  = StandardScaler()
sc_x2  = StandardScaler()
sc_x3  = StandardScaler()
sc_x4  = StandardScaler()
sc_x5  = StandardScaler()
sc_x6  = StandardScaler()
sc_x7  = StandardScaler()
sc_x10 = StandardScaler()

# Ajuste en TRAIN y transformación en TRAIN/TEST (columna a columna)
esc3_train['x1']  = sc_x1.fit_transform( esc2_train[['x1']] ).ravel()
esc3_test['x1']   = sc_x1.transform(      esc2_test[['x1']]  ).ravel()

esc3_train['x2']  = sc_x2.fit_transform( esc2_train[['x2']] ).ravel()
esc3_test['x2']   = sc_x2.transform(      esc2_test[['x2']]  ).ravel()

esc3_train['x3']  = sc_x3.fit_transform( esc2_train[['x3']] ).ravel()
esc3_test['x3']   = sc_x3.transform(      esc2_test[['x3']]  ).ravel()

esc3_train['x4']  = sc_x4.fit_transform( esc2_train[['x4']] ).ravel()
esc3_test['x4']   = sc_x4.transform(      esc2_test[['x4']]  ).ravel()

esc3_train['x5']  = sc_x5.fit_transform( esc2_train[['x5']] ).ravel()
esc3_test['x5']   = sc_x5.transform(      esc2_test[['x5']]  ).ravel()

esc3_train['x6']  = sc_x6.fit_transform( esc2_train[['x6']] ).ravel()
esc3_test['x6']   = sc_x6.transform(      esc2_test[['x6']]  ).ravel()

esc3_train['x7']  = sc_x7.fit_transform( esc2_train[['x7']] ).ravel()
esc3_test['x7']   = sc_x7.transform(      esc2_test[['x7']]  ).ravel()

esc3_train['x10'] = sc_x10.fit_transform(esc2_train[['x10']]).ravel()
esc3_test['x10']  = sc_x10.transform(     esc2_test[['x10']] ).ravel()

# NOTA: x8 NO se escala (es categórica y la trataremos en Escenario 4 con dummies)

# ---- Reporte de escalado para exposición (media y escala aprendidas en TRAIN) ----
scalers = {
    'x1': sc_x1, 'x2': sc_x2, 'x3': sc_x3, 'x4': sc_x4,
    'x5': sc_x5, 'x6': sc_x6, 'x7': sc_x7, 'x10': sc_x10
}
print("Parámetros de escalado (aprendidos en TRAIN):")
for col, sc in scalers.items():
    print(f"  {col}: mean_={sc.mean_[0]:.6f} | scale_={sc.scale_[0]:.6f}")

# ---- Chequeo: en TRAIN las continuas deben quedar ~N(0,1) ----
_cols_cont = ['x1','x2','x3','x4','x5','x6','x7','x10']
max_abs_mean = float(np.abs(esc3_train[_cols_cont].mean()).max())
max_std_dev  = float(np.abs(esc3_train[_cols_cont].std(ddof=0) - 1).max())  # ddof=0 para ser coherentes con StandardScaler
print(f"\nChequeo TRAIN post-escalado → max|media|={max_abs_mean:.4e}, max|std-1|={max_std_dev:.4e}")

# ---- Auditoría (NaN permitidos solo en x8; el resto deben estar limpios) ----
audit_pair(esc3_train, esc3_test, "Escenario 3", allow_nans_cols=['x8'])
```

```python
# === Predicción con ensemble (input = DataFrame ya en Esc6) ===
import json, joblib, numpy as np, pandas as pd, xgboost as xgb
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

def predict_ensemble_from_esc6(df_esc6: pd.DataFrame, y_true=None, arte_dir="artefactos_simple"):
    # Cargar artefactos
    with open(f"{arte_dir}/ensemble.json","r",encoding="utf-8") as f:
        meta = json.load(f)
    alpha = float(meta["alpha_xgb"])
    cols  = meta["cols_train"]

    booster = xgb.Booster(); booster.load_model(f"{arte_dir}/xgb.json")
    rf      = joblib.load(f"{arte_dir}/rf.pkl")

    # Alinear columnas
    X = df_esc6.reindex(columns=cols, fill_value=0)

    # Predicción XGB
    dmat = xgb.DMatrix(X)
    try:
        yhat_xgb = booster.predict(dmat, iteration_range=(0, booster.best_iteration+1))
    except Exception:
        yhat_xgb = booster.predict(dmat)

    # Predicción RF
    yhat_rf = rf.predict(X)

    # Ensemble
    yhat = alpha * yhat_xgb + (1 - alpha) * yhat_rf

    # Métricas opcionales si pasas y_true
    if y_true is not None:
        rmse = float(np.sqrt(mean_squared_error(y_true, yhat)))
        mae  = float(mean_absolute_error(y_true, yhat))
        r2   = float(r2_score(y_true, yhat))
        print(f"[Ensemble] RMSE={rmse:.4f} | MAE={mae:.4f} | R2={r2:.4f}")
    return yhat

# --- EJEMPLO: evaluar en tu hold-out actual (esc6_test / y_test) ---
_ = predict_ensemble_from_esc6(esc6_test, y_true=y_test, arte_dir="artefactos_simple")
```

```python
# === Predicción FINAL sobre data/test_inputs.* con el ensemble simple (explicable) ===


# -------------------------
# 1) Utilidades (claritas)
# -------------------------
def to_numeric_series(s: pd.Series) -> pd.Series:
    """Convierte strings a números (captura dígitos) y fuerza NaN si no hay número."""
    if pd.api.types.is_numeric_dtype(s):
        return pd.to_numeric(s, errors="coerce")
    return pd.to_numeric(
        s.astype(str).str.extract(r"(-?\d+(?:\.\d+)?)", expand=False),
        errors="coerce"
    )

def engineer_feats(df: pd.DataFrame) -> pd.DataFrame:
    """Crea x6/x7, x6-x7 y log1p(x8) de forma segura y limpia infinitos."""
    out = df.copy()
    for c in ["x6","x7","x8"]:
        if c in out: out[c] = pd.to_numeric(out[c], errors="coerce")
    if {"x6","x7"}.issubset(out.columns):
        denom = out["x7"].replace(0, np.nan)
        out["x6_div_x7"]   = out["x6"] / denom
        out["x6_minus_x7"] = out["x6"] - out["x7"]
    if "x8" in out.columns:
        out["x8_log1p"] = np.log1p(np.clip(out["x8"], a_min=0, a_max=None))
    return out.replace([np.inf, -np.inf], np.nan)

def load_table_guess(base: Path) -> pd.DataFrame:
    """Intenta cargar test_inputs desde .xlsx o .csv en /data o raíz."""
    candidates = [
        base / "data" / "test_inputs.xlsx",
        base / "data" / "test_inputs.csv",
        base / "test_inputs.xlsx",
        base / "test_inputs.csv",
    ]
    for p in candidates:
        if p.exists():
            return pd.read_excel(p) if p.suffix.lower() in [".xlsx", ".xls"] else pd.read_csv(p)
    raise FileNotFoundError("No encontré test_inputs (.xlsx/.csv) en ./data o raíz.")

# --------------------------------------------------------
# 2) Aprender estadísticas de TRAIN (para replicar Esc6)
#    (winsor → features → imputación → Yeo-Johnson)
# --------------------------------------------------------
BASE = Path(".")
train_path = (BASE / "data" / "train.xlsx") if (BASE / "data" / "train.xlsx").exists() else (BASE / "train.xlsx")
train_df = pd.read_excel(train_path)

# Separamos y tipamos lo básico
train_X = train_df.drop(columns=["deseada"]).copy()

# a) x8 numérica y resto a numérico
for c in ['x1','x2','x3','x4','x5','x6','x7','x10']:
    if c in train_X: train_X[c] = pd.to_numeric(train_X[c], errors='coerce')
if 'x8' in train_X: train_X['x8'] = to_numeric_series(train_X['x8'])

# b) WINSOR (solo en train, y guardamos límites)
winsor_cols = [c for c in ['x1','x2','x5','x8'] if c in train_X.columns]
q_lo = train_X[winsor_cols].quantile(0.005)
q_hi = train_X[winsor_cols].quantile(0.995)
train_X[winsor_cols] = train_X[winsor_cols].clip(lower=q_lo, upper=q_hi, axis=1)

# c) FEATURES
train_X = engineer_feats(train_X)

# d) IMPUTACIÓN (medianas en train)
impute_cols = [c for c in ['x1','x2','x3','x4','x5','x6','x7','x10','x8',
                           'x6_div_x7','x6_minus_x7','x8_log1p'] if c in train_X.columns]
medianas = train_X[impute_cols].median(numeric_only=True)
train_X[impute_cols] = train_X[impute_cols].fillna(medianas)

# e) YEO-JOHNSON (aprendemos λ en train y lo reutilizamos)
yj_cols = [c for c in ['x1','x2','x3','x5','x8'] if c in train_X.columns]
lambdas_yj = {}
for c in yj_cols:
    train_X[c], lam = yeojohnson(train_X[c].values)  # devuelve transformado + λ
    lambdas_yj[c] = float(lam)

# --------------------------------------------------------
# 3) Transformar TEST_INPUTS con esas estadísticas
# --------------------------------------------------------
test_raw = load_table_guess(BASE)

# Tipado básico + x8 numérica
test_X = test_raw.copy()
for c in ['x1','x2','x3','x4','x5','x6','x7','x10']:
    if c in test_X: test_X[c] = pd.to_numeric(test_X[c], errors='coerce')
if 'x8' in test_X: test_X['x8'] = to_numeric_series(test_X['x8'])

# Winsor con límites de TRAIN
for c in winsor_cols:
    if c in test_X:
        test_X[c] = pd.to_numeric(test_X[c], errors="coerce").clip(lower=q_lo[c], upper=q_hi[c])

# Features
test_X = engineer_feats(test_X)

# Imputación con medianas de TRAIN
for c in impute_cols:
    if c not in test_X: test_X[c] = np.nan
test_X[impute_cols] = test_X[impute_cols].fillna(medianas)

# Yeo-Johnson con los λ de TRAIN
for c, lam in lambdas_yj.items():
    if c in test_X:
        test_X[c] = yeojohnson(test_X[c].values, lmbda=lam)

# --------------------------------------------------------
# 4) Cargar ensemble y predecir
# --------------------------------------------------------
# Cargamos α, columnas y modelos desde artefactos_simple
with open("artefactos_simple/ensemble.json","r",encoding="utf-8") as f:
    meta = json.load(f)
alpha = float(meta["alpha_xgb"])
cols  = meta["cols_train"]

booster = xgb.Booster(); booster.load_model("artefactos_simple/xgb.json")
rf      = joblib.load("artefactos_simple/rf.pkl")

# Alinear test a las columnas de entrenamiento (Esc6)
X_test_esc6 = test_X.reindex(columns=cols, fill_value=0)

# Predicciones XGB (usando best_iteration si existe en el booster)
dte = xgb.DMatrix(X_test_esc6)
try:
    yhat_xgb = booster.predict(dte, iteration_range=(0, booster.best_iteration+1))
except Exception:
    yhat_xgb = booster.predict(dte)

# Predicción RF
yhat_rf = rf.predict(X_test_esc6)

# Ensemble
yhat = alpha * yhat_xgb + (1 - alpha) * yhat_rf

# --------------------------------------------------------
# 5) Guardar envío
# --------------------------------------------------------
out = test_raw.copy()
out["pred_deseada"] = yhat
out_path = "predicciones_ensemble_submit.csv"
out.to_csv(out_path, index=False, encoding="utf-8")
print(f"✅ Predicciones guardadas en {out_path}")
print(out.head(3))
```

```python
# === Eval + Gráficas (autocontenida y simple) ===

def rmse(a,b): return float(np.sqrt(mean_squared_error(a,b)))

# --- 0) Cargar artefactos del ensemble simple ---
with open("artefactos_simple/ensemble.json","r",encoding="utf-8") as f:
    meta = json.load(f)
alpha = float(meta["alpha_xgb"])
cols  = meta["cols_train"]

booster = xgb.Booster(); booster.load_model("artefactos_simple/xgb.json")
rf      = joblib.load("artefactos_simple/rf.pkl")

# --- 1) Fuente de datos para evaluación ---
if "esc6_test" in globals() and "y_test" in globals():
    X_src = esc6_test.copy()
    y_src = y_test.copy()
else:
    raise RuntimeError("Necesito 'esc6_test' y 'y_test' en memoria (formato Esc6). Ejecútalos antes.")

# Alinear columnas al orden de entrenamiento
X_eval = X_src.reindex(columns=cols, fill_value=0)

# --- 2) Predicciones (XGB, RF y ensemble) ---
dmat = xgb.DMatrix(X_eval)
try:
    yhat_xgb = booster.predict(dmat, iteration_range=(0, booster.best_iteration+1))
except Exception:
    yhat_xgb = booster.predict(dmat)

yhat_rf  = rf.predict(X_eval)
yhat_ens = alpha * yhat_xgb + (1 - alpha) * yhat_rf

# --- 3) Métricas y eval_df ---
m_rmse = rmse(y_src, yhat_ens)
m_mae  = float(mean_absolute_error(y_src, yhat_ens))
m_r2   = float(r2_score(y_src, yhat_ens))
print(f"[Ensemble] TEST → RMSE={m_rmse:.4f} | MAE={m_mae:.4f} | R2={m_r2:.4f} | α={alpha:.3f}")

eval_df = pd.DataFrame({
    "y": y_src.values,
    "yhat_ens": yhat_ens,
    "yhat_xgb": yhat_xgb,
    "yhat_rf":  yhat_rf,
})
eval_df["residual"]  = eval_df["y"] - eval_df["yhat_ens"]
eval_df["abs_error"] = eval_df["residual"].abs()

# --- 4) Gráficas de diagnóstico básicas ---
# 4.1 y vs ŷ (calibración)
plt.figure(figsize=(5.5,5))
plt.scatter(eval_df["y"], eval_df["yhat_ens"], s=14, alpha=0.5)
m = [min(eval_df["y"].min(), eval_df["yhat_ens"].min()),
     max(eval_df["y"].max(), eval_df["yhat_ens"].max())]
plt.plot(m, m, lw=2)
plt.title("Calibración: y real vs ŷ (ensemble)")
plt.xlabel("y real"); plt.ylabel("ŷ predicha")
plt.tight_layout()
plt.show()

# 4.2 Histograma de residuales
plt.figure(figsize=(5.5,4))
plt.hist(eval_df["residual"], bins=30)
plt.title("Distribución de residuales (y − ŷ)")
plt.xlabel("residual"); plt.ylabel("frecuencia")
plt.tight_layout()
plt.show()

# 4.3 Residual vs ŷ (heteroscedasticidad)
plt.figure(figsize=(5.5,4))
plt.scatter(eval_df["yhat_ens"], eval_df["residual"], s=14, alpha=0.5)
plt.axhline(0, lw=2)
plt.title("Residual vs ŷ (ensemble)")
plt.xlabel("ŷ predicha"); plt.ylabel("residual")
plt.tight_layout()
plt.show()

# 4.4 MAE por deciles de y
eval_df["y_decile"] = pd.qcut(eval_df["y"], 10, labels=False, duplicates="drop")
mae_by_decile = eval_df.groupby("y_decile")["abs_error"].mean()
plt.figure(figsize=(6,4))
plt.bar(range(len(mae_by_decile)), mae_by_decile.values)
plt.title("MAE medio por decil de y")
plt.xlabel("decil de y (bajo → alto)"); plt.ylabel("MAE medio")
plt.tight_layout()
plt.show()

# (Opcional) Top-errores para inspección rápida
display(eval_df.sort_values("abs_error", ascending=False).head(10))
```
