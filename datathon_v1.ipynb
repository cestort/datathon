### Gráfico de distribución para cada variable numérica


```python
# Repasar los comentarios del código y cambiarlos a mejoras expositivas

import math  # Para calcular el número de filas necesarias


# --- Selección de variables y dimensionado del lienzo ---

columnas_numeric = df.select_dtypes(include=np.number).columns
n_vars = len(columnas_numeric)
if n_vars == 0:
    raise ValueError("El DataFrame no contiene columnas numéricas.")

# Máximo de columnas de subplots por fila (ajustable cambiando esta variable)
ncols = 3
# Filas necesarias para cubrir todas las variables
nrows = math.ceil(n_vars / ncols)

fig, axes = plt.subplots(nrows=nrows, ncols=ncols,
                         figsize=(3*ncols, 2.5*nrows))

# Convertir a array 1D para indexar sin errores, incluso si hay una sola fila/columna
axes = np.atleast_1d(axes).ravel()

# Paleta de colores (rotará si hay más variables que colores)
palette = sns.color_palette(n_colors=max(n_vars, 6))

# --- Bucle de trazado: un histograma por variable numérica ---
for i, col in enumerate(columnas_numeric):
    ax = axes[i]
    sns.histplot(
        data=df, x=col, stat="count", kde=True,
        color=palette[i % len(palette)], ax=ax
    )
    ax.set_title(col, fontsize=10)
    ax.tick_params(labelsize=8)
    ax.set_xlabel("")

# --- Ocultar ejes no usados (si la rejilla quedó más grande de lo necesario) ---
for ax in axes[n_vars:]:
    ax.set_visible(False)

# --- Ajustes finales de maquetación ---
fig.tight_layout()
plt.subplots_adjust(top=0.9)
fig.suptitle('Distribución de variables numéricas', fontsize=12)
```

Según las gráficas , las variables x4, x6, x7 y x9 tienen las mejores distribuciones. La columna a predecir [deseada] tiene una buena distribución.

```python
# Normalidad en la variable respuesta "deseada"

shapiro_test = stats.shapiro(df["deseada"])
print(
    f"Test Shapiro-Wilk: estadístico = {shapiro_test[0]}, p-value = {shapiro_test[1]}")
```

```python
def apply_boxcox(series, median, lam):
    s = series.fillna(median).astype(float).to_numpy()
    return pd.Series(boxcox(s, lmbda=lam), index=series.index, name=series.name)

def apply_yeojohnson(series, median, lam):
    s = series.fillna(median).astype(float).to_numpy()
    return pd.Series(yeojohnson(s, lmbda=lam), index=series.index, name=series.name)

# Construye TEST transformado usando tus medianas y lambdas de TRAIN (sin recomputar nada)
X_test_t = pd.DataFrame({
    'x1':  apply_boxcox(    X_test['x1'],  x1_median,  x1_lambda),
    'x2':  apply_yeojohnson(X_test['x2'],  x2_median,  x2_lambda),
    'x3':  apply_yeojohnson(X_test['x3'],  x3_median,  x3_lambda),
    'x4':  X_test['x4'].fillna(x4_median).astype(float),
    'x5':  apply_yeojohnson(X_test['x5'],  x5_median,  x5_lambda),
    'x6':  X_test['x6'].fillna(x6_median).astype(float),
    'x7':  X_test['x7'].fillna(x7_median).astype(float),
    'x8':  apply_boxcox(    X_test['x8'],  x8_median,  x8_lambda),
    # 'x9': (omitida si no la usas)
    'x10': X_test['x10'].fillna(x10_median).astype(float),
}, index=X_test.index)[X_train_t.columns]  # mismo orden que TRAIN
```

```python

def audit_shapes(df, Xtr, Xte, ytr, yte, name="dataset"):
    print(f"\n[{name}]")
    print(f"  df:        {df.shape}")
    print(f"  X_train:   {Xtr.shape} | X_test:  {Xte.shape}")
    print(f"  y_train:   {ytr.shape} | y_test:  {yte.shape}")
    assert len(Xtr)==len(ytr) and len(Xte)==len(yte), "Tamaños X/y incoherentes"

def audit_scenario_pair(Xtr, Xte, name):
    print(f"\n[Escenario {name}]")
    # target no presente
    bad_cols = [c for c in Xtr.columns if c.lower() in ("deseada","target","y")]
    assert not bad_cols, f"Target fugada en {name}: {bad_cols}"
    # columnas iguales
    only_train = sorted(set(Xtr.columns)-set(Xte.columns))
    only_test  = sorted(set(Xte.columns)-set(Xtr.columns))
    if only_train or only_test:
        print("  ⚠ columnas desalineadas (se corregirá con reindex en TEST):")
        if only_train: print("   - Solo en train:", only_train[:10], "..." if len(only_train)>10 else "")
        if only_test:  print("   - Solo en test :", only_test[:10], "..." if len(only_test)>10 else "")
    else:
        print("  ✓ columnas alineadas")

    # NaNs/inf
    nans_tr = np.isnan(Xtr.to_numpy(dtype=float)).sum()
    nans_te = np.isnan(Xte.to_numpy(dtype=float)).sum()
    infs_tr = np.isinf(Xtr.to_numpy(dtype=float)).sum()
    infs_te = np.isinf(Xte.to_numpy(dtype=float)).sum()
    print(f"  NaN train/test: {nans_tr}/{nans_te} | inf train/test: {infs_tr}/{infs_te}")

# ---- ejecuta auditoría ----
audit_shapes(df, X_train, X_test, y_train, y_test, name="principal")

ESC_TR = {
    "escenario1": escenario1_train, "escenario2": escenario2_train, "escenario3": escenario3_train,
    "escenario4": escenario4_train, "escenario5": escenario5_train,
}
ESC_TE = {
    "escenario1": escenario1_test,  "escenario2": escenario2_test,  "escenario3": escenario3_test,
    "escenario4": escenario4_test,  "escenario5": escenario5_test,
}

for k in ESC_TR:
    audit_scenario_pair(ESC_TR[k], ESC_TE[k], k)

print("\nAuditoría completada. Si hay advertencias de columnas desalineadas, el bloque de evaluación reindexará TEST a TRAIN.")
```

```python
# =================== TUNING Y EVALUACIÓN — XGB (Esc1) y GB (Esc2) ===================
# Prioridad: simplicidad y claridad
import numpy as np
import pandas as pd
from IPython.display import display

from sklearn.model_selection import RandomizedSearchCV, KFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, make_scorer
from sklearn.ensemble import GradientBoostingRegressor

# -------------------------------------------------------------------
# 1) Utilidades mínimas
# -------------------------------------------------------------------
def prepare_xy(X_train_df, X_test_df, y_train, y_test):
    """Alinea índices/columnas, limpia infinitos y devuelve X_tr, X_te, y_tr, y_te."""
    y_tr = y_train.reindex(X_train_df.index)
    y_te = y_test.reindex(X_test_df.index)
    X_te = X_test_df.reindex(columns=X_train_df.columns, fill_value=0)
    X_tr = X_train_df.replace([np.inf, -np.inf], np.nan)
    X_te = X_te.replace([np.inf, -np.inf], np.nan)
    return X_tr, X_te, y_tr, y_te

def make_scoring():
    """Multi-métrica: minimizar RMSE y reportar R2 en CV."""
    rmse = make_scorer(lambda yt, yp: np.sqrt(mean_squared_error(yt, yp)),
                       greater_is_better=False)
    r2   = make_scorer(r2_score)
    return {"rmse": rmse, "r2": r2}

def show_cv_top(cv_results, top=5, title="TOP (CV)"):
    """Ordena por mejor RMSE (rank_test_rmse) y muestra tabla compacta."""
    df = pd.DataFrame(cv_results).sort_values("rank_test_rmse").reset_index(drop=True)
    df["RMSE_CV"] = -df["mean_test_rmse"]      # convertir a positivo
    df["R2_CV"]   =  df["mean_test_r2"]
    cols = ["params", "RMSE_CV", "R2_CV", "std_test_rmse", "std_test_r2"]
    print(f"\n{title}")
    display(df[cols].head(top))
    return df  # por si quieres guardarlo

def report_test(y_true, y_pred, modelo=""):
    rmse = float(np.sqrt(mean_squared_error(y_true, y_pred)))
    mae  = float(mean_absolute_error(y_true, y_pred))
    r2   = float(r2_score(y_true, y_pred))
    print(f"→ TEST {modelo}: RMSE={rmse:.4f} | MAE={mae:.4f} | R2={r2:.4f}")
    return {"RMSE": rmse, "MAE": mae, "R2": r2}

# CV común y scoring
cv5 = KFold(n_splits=5, shuffle=True, random_state=42)
scoring = make_scoring()

# -------------------------------------------------------------------
# 2) Opción A — Escenario 1 + XGBoost
# -------------------------------------------------------------------
print("\n========== Opción A: Escenario 1 + XGBoost ==========")
try:
    from xgboost import XGBRegressor

    # Datos
    Xtr_A, Xte_A, ytr_A, yte_A = prepare_xy(escenario1_train, escenario1_test, y_train, y_test)

    # Pipeline simple: Imputación + XGB
    pipe_xgb = Pipeline([
        ("imp", SimpleImputer(strategy="median")),
        ("est", XGBRegressor(
            objective="reg:squarederror",
            tree_method="hist",
            n_jobs=-1,
            random_state=42
        ))
    ])

    # Espacio de búsqueda (sencillo y efectivo)
    param_dist_A = {
        "est__n_estimators":     [800, 1000, 1200, 1400, 1500],
        "est__learning_rate":    [0.03, 0.04, 0.05, 0.06, 0.07],
        "est__max_depth":        [4, 5, 6],
        "est__subsample":        [0.7, 0.75, 0.8, 0.85, 0.9],
        "est__colsample_bytree": [0.6, 0.7, 0.8, 0.9],
        "est__reg_lambda":       [1.0, 1.5, 2.0, 3.0, 5.0],
    }

    rs_xgb = RandomizedSearchCV(
        estimator=pipe_xgb,
        param_distributions=param_dist_A,
        n_iter=40,
        scoring=scoring,   # RMSE + R2
        refit="rmse",      # elige el mejor por RMSE (CV)
        cv=cv5,
        n_jobs=-1,
        random_state=42,
        verbose=0,
        return_train_score=False
    )
    rs_xgb.fit(Xtr_A, ytr_A)

    # CV Top y TEST
    cv_xgb_df = show_cv_top(rs_xgb.cv_results_, title="TOP XGB (CV)")
    best_xgb  = rs_xgb.best_estimator_
    ypred_A   = best_xgb.predict(Xte_A)
    met_A     = report_test(yte_A, ypred_A, "XGB")
    print(f"Mejor XGB (CV RMSE): {-rs_xgb.best_score_:.4f} | Params: {rs_xgb.best_params_}")

except Exception as e:
    print("XGBoost no disponible o error durante el tuning:", repr(e))
    best_xgb, cv_xgb_df, met_A = None, None, None

# -------------------------------------------------------------------
# 3) Opción B — Escenario 2 + Gradient Boosting
# -------------------------------------------------------------------
print("\n========== Opción B: Escenario 2 + Gradient Boosting ==========")

# Datos
Xtr_B, Xte_B, ytr_B, yte_B = prepare_xy(escenario2_train, escenario2_test, y_train, y_test)

# Pipeline simple: Imputación + GB
pipe_gb = Pipeline([
    ("imp", SimpleImputer(strategy="median")),
    ("est", GradientBoostingRegressor(random_state=42))
])

param_dist_B = {
    "est__n_estimators":  [400, 600, 800, 1000, 1200],
    "est__learning_rate": [0.03, 0.04, 0.05, 0.06, 0.08],
    "est__max_depth":     [2, 3, 4],
    "est__subsample":     [0.6, 0.7, 0.8, 0.9, 1.0],
}

rs_gb = RandomizedSearchCV(
    estimator=pipe_gb,
    param_distributions=param_dist_B,
    n_iter=40,
    scoring=scoring,   # RMSE + R2
    refit="rmse",
    cv=cv5,
    n_jobs=-1,
    random_state=42,
    verbose=0,
    return_train_score=False
)
rs_gb.fit(Xtr_B, ytr_B)

cv_gb_df = show_cv_top(rs_gb.cv_results_, title="TOP GB (CV)")
best_gb  = rs_gb.best_estimator_
ypred_B  = best_gb.predict(Xte_B)
met_B    = report_test(yte_B, ypred_B, "GB")
print(f"Mejor GB (CV RMSE): {-rs_gb.best_score_:.4f} | Params: {rs_gb.best_params_}")

# -------------------------------------------------------------------
# 4) Resumen final (sólo lo esencial)
# -------------------------------------------------------------------
print("\n================== RESUMEN ==================")
if met_A:
    print(f"[Esc1 – XGB]  TEST  | RMSE={met_A['RMSE']:.4f} | MAE={met_A['MAE']:.4f} | R2={met_A['R2']:.4f}")
print(f"[Esc2 –  GB]  TEST  | RMSE={met_B['RMSE']:.4f} | MAE={met_B['MAE']:.4f} | R2={met_B['R2']:.4f}")


```
